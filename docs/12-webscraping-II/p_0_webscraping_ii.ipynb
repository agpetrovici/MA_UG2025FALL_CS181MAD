````python
r = requests.get("https://usc-webcomponents.toyota-europe.com/v1/api/usedcars/results/es/es?brand=toyota")

r.url
````

````python
params = {
    "brand": "toyota",
    "foo": "bar",
}
r = requests.get(
    "https://usc-webcomponents.toyota-europe.com/v1/api/usedcars/results/es/es",
    params=params,
)

r.url
````
